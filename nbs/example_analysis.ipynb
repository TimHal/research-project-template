# Example Analysis Notebook

This notebook demonstrates how to load trained models and analyze results.

```python
import sys
sys.path.insert(0, '../src')

import torch
import matplotlib.pyplot as plt
```

## Load Model from Config

```python
from util.model_loading import load_from_config

# Load model and datamodule from config (no checkpoint)
model, dm = load_from_config('../conf/experiment/cifar10_classification.yaml')

# Setup data
dm.setup('fit')

print(f'Model: {model}')
print(f'Dataset: {dm.dataset_name}')
print(f'Train samples: {len(dm.train_dataset)}')
print(f'Val samples: {len(dm.val_dataset)}')
```

## Visualize Training Data

```python
from util.viz_utils import batch_to_grid

# Get a batch of training data
batch, labels = next(iter(dm.train_dataloader()))

# Display as grid
grid = batch_to_grid(
    batch[:16], 
    nrow=4,
    mean=dm.norm_mean,
    std=dm.norm_std
)

plt.figure(figsize=(10, 10))
plt.imshow(grid)
plt.axis('off')
plt.title('Sample Training Images')
plt.show()
```

## Model Inference

```python
# Run inference on a batch
model.eval()
with torch.no_grad():
    logits = model(batch[:8])
    predictions = logits.argmax(dim=1)

print('Predictions:', predictions.tolist())
print('Ground truth:', labels[:8].tolist())
```
